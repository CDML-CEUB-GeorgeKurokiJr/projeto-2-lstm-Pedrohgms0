### Hipótese de trabalho

A modelagem conjunta de ativos correlacionados verticalmente na cadeia de exploração de petróleo — combinada com features técnicas, macroeconômicas e função de perda híbrida — permite capturar relações não-lineares entre variáveis que uma estratégia passiva (buy & hold) não explora. A hipótese operacional é que o modelo gera alpha estatisticamente positivo em período out-of-sample estritamente separado do treino.

````python
#!pip install torch yfinance pandas-ta scikit-learn seaborn matplotlib scipy

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

import yfinance as yf
import pandas_ta as ta

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# ── Reprodutibilidade ─────────────────────────────────────────────────────────
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device: {DEVICE}  |  PyTorch: {torch.__version__}  |  pandas-ta: {ta.version}')

# ── Helper: detecta nomes de colunas do pandas-ta independente da versão ──────
def get_col(df, prefix):
    cols = [c for c in df.columns if str(c).startswith(prefix)]
    if not cols:
        raise KeyError(f"Prefixo '{prefix}' não encontrado. Disponíveis: {list(df.columns)}")
    return cols[0]

plt.rcParams.update({
    'figure.facecolor': 'white', 'axes.facecolor': 'white',
    'axes.grid': True, 'grid.alpha': 0.3, 'font.size': 11,
})
CORES = {'XOM': '#1D9E75', 'CVX': '#534AB7', 'SLB': '#D85A30', 'HAL': '#BA7517'}
print('OK')
````

## 2. Coleta de Dados

Fonte primária: `yfinance` (preços OHLCV ajustados por dividendos e splits).  
Fontes macro: contratos futuros de WTI (`CL=F`), Brent (`BZ=F`) e gás natural (`NG=F`).

**Alinhamento temporal:** feriados e dias sem pregão são preenchidos por *forward-fill* (último valor disponível propagado para frente). Isso é uma aproximação: implica assumir que o preço não mudou, o que pode subestimar a volatilidade em torno de feriados. Valores ausentes remanescentes após o alinhamento são removidos.

> O período 2020–2022 é excluído **antes** de qualquer ajuste de scaler ou criação de janelas, garantindo que nenhuma informação desse intervalo contamine o pipeline.


````python
TICKERS_ATIVOS = ['XOM', 'CVX', 'SLB', 'HAL']
TICKERS_MACRO  = ['CL=F', 'BZ=F', 'NG=F']

# 2020-2022 excluído — pandemia
PERIODOS = [
    ('2010-01-01', '2019-12-31'),
    ('2023-01-01', '2025-12-31'),
]

def baixar_concat(tickers, periodos):
    partes = [yf.download(tickers, start=s, end=e,
                          auto_adjust=True, progress=False)
              for s, e in periodos]
    return pd.concat(partes).sort_index()

raw_ativos = baixar_concat(TICKERS_ATIVOS, PERIODOS)
raw_macro  = baixar_concat(TICKERS_MACRO,  PERIODOS)

print(f'Total: {len(raw_ativos)} dias')
print(f'Treino (2010-2019): {len(raw_ativos[raw_ativos.index < "2020-01-01"])} dias')
print(f'Teste  (2023-2025): {len(raw_ativos[raw_ativos.index >= "2023-01-01"])} dias')
raw_ativos['Close'].tail(3)
````

## 3. Engenharia de Features

**13 features em 5 grupos** — selecionadas com base em três critérios:

1. **Estacionariedade:** todas as features de preço são transformadas em log-retorno. A hipótese de estacionariedade foi verificada formalmente pelo Teste ADF (p < 0,05 para todas as séries transformadas). Detalhes em `exploratory_analysis.ipynb`.
2. **Ausência de redundância linear:** pares com |r| > 0,90 foram eliminados (ver análise completa no notebook exploratório).
3. **Poder preditivo:** Mutual Information calculada entre cada feature e o log-retorno do dia seguinte — apenas features com MI > 0 foram mantidas.

> **Nota metodológica:** a matriz de correlação e o MI foram calculados **somente no conjunto de treino (2010–2019)**. Aplicar esses critérios sobre o conjunto completo (incluindo o teste) constituiria *data leakage* por seleção de features.


````python
FEATURES = [
    'logret_close', 'logret_open',
    'volume_ratio',
    'sma_diff', 'ema_diff',
    'rsi14', 'macd_hist',
    'atr14', 'bb_pct_b',
    'wti_logret', 'brent_logret', 'spread_wb', 'ng_logret',
]

def calcular_features(ticker, raw_ativos, raw_macro):
    df     = pd.DataFrame(index=raw_ativos.index)
    close  = raw_ativos['Close'][ticker]
    open_  = raw_ativos['Open'][ticker]
    high   = raw_ativos['High'][ticker]
    low    = raw_ativos['Low'][ticker]
    volume = raw_ativos['Volume'][ticker]

    df['logret_close'] = np.log(close / close.shift(1))
    df['logret_open']  = np.log(open_ / close.shift(1))
    df['volume_ratio'] = volume / volume.rolling(20).mean()

    sma50  = ta.sma(close, length=50)
    sma200 = ta.sma(close, length=200)
    ema12  = ta.ema(close, length=12)
    ema26  = ta.ema(close, length=26)
    df['sma_diff'] = (sma50 - sma200) / sma200
    df['ema_diff'] = (ema12 - ema26)  / close

    df['rsi14']     = ta.rsi(close, length=14)
    macd_df         = ta.macd(close, fast=12, slow=26, signal=9)
    df['macd_hist'] = macd_df[get_col(macd_df, 'MACDh')] / close

    df['atr14']    = ta.atr(high, low, close, length=14) / close
    bb             = ta.bbands(close, length=20, std=2)
    df['bb_pct_b'] = bb[get_col(bb, 'BBP')]

    wti = np.log(raw_macro['Close']['CL=F'] / raw_macro['Close']['CL=F'].shift(1))
    brt = np.log(raw_macro['Close']['BZ=F'] / raw_macro['Close']['BZ=F'].shift(1))
    ng  = np.log(raw_macro['Close']['NG=F'] / raw_macro['Close']['NG=F'].shift(1))
    df['wti_logret']   = wti
    df['brent_logret'] = brt
    df['ng_logret']    = ng
    df['spread_wb']    = wti - brt

    df['target'] = df['logret_close'].shift(-1)
    return df.dropna()


features_ativos = {t: calcular_features(t, raw_ativos, raw_macro)
                   for t in TICKERS_ATIVOS}

for t, df in features_ativos.items():
    n_tr = len(df[df.index < '2020-01-01'])
    n_te = len(df[df.index >= '2023-01-01'])
    print(f'{t}:  treino={n_tr}  teste={n_te}')
````

## 4. Pipeline de Dados

**Split temporal (estritamente cronológico; sem embaralhamento):**

- Treino: 2010–2019 (~2500 dias)
- Teste out-of-sample: 2023–2025 (~600 dias)

**Normalização:** `StandardScaler` ajustado **exclusivamente** nos dados de treino e aplicado ao teste sem reajuste. Reajustar o scaler no teste constituiria *data leakage* temporal: o modelo teria acesso, de forma implícita, a estatísticas do futuro.

**Janelas deslizantes:** lookback de 20 dias, stride=1. Cada amostra é um tensor de forma `(20, 13)` com o alvo sendo o log-retorno do dia `t+1`.

> **Limitação:** o protocolo atual não possui conjunto de validação independente. O early stopping da Seção 6 monitora a loss de **treino**, não de validação — o que reduz sua eficácia como regularizador. Uma partição temporal de validação (ex.: 2017–2019) seria a alternativa metodologicamente correta; a ausência dela é um ponto fraco reconhecido desta versão.


````python
LOOKBACK = 20
BATCH    = 64

# ── Constantes de capital (usadas na Seção 9 e na robustez) ──────────────────
CAPITAL_INICIAL   = 10_000          # capital total inicial ($)
CAPITAL_POR_ATIVO = CAPITAL_INICIAL / 4   # igual para os 4 ativos

def simular_capital(retornos_log, capital_inicial):
    """Converte log-retornos em curva de capital com capitalização contínua."""
    return capital_inicial * np.exp(np.cumsum(retornos_log))

def criar_janelas(X, y, lookback):
    Xs, ys = [], []
    for i in range(lookback, len(X)):
        Xs.append(X[i-lookback:i])
        ys.append(y[i])
    return np.array(Xs, dtype=np.float32), np.array(ys, dtype=np.float32)

class SerieDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.from_numpy(X)
        self.y = torch.from_numpy(y)
    def __len__(self):        return len(self.X)
    def __getitem__(self, i): return self.X[i], self.y[i]


def preparar_pipeline(features_ativos, tickers, lookback, batch):
    scalers  = {}
    janelas  = {'treino': {}, 'teste': {}}

    for t in tickers:
        df = features_ativos[t]
        tr = df[df.index <  '2020-01-01']
        te = df[df.index >= '2023-01-01']

        sc = StandardScaler()
        X_tr = sc.fit_transform(tr[FEATURES].values)
        X_te = sc.transform(te[FEATURES].values)
        scalers[t] = sc

        janelas['treino'][t] = criar_janelas(X_tr, tr['target'].values, lookback)
        janelas['teste'][t]  = criar_janelas(X_te, te['target'].values, lookback)

    def concat(split):
        Xs = np.concatenate([janelas[split][t][0] for t in tickers])
        ys = np.concatenate([janelas[split][t][1] for t in tickers])
        return Xs, ys

    X_tr, y_tr = concat('treino')
    X_te, y_te = concat('teste')

    loader_treino = DataLoader(SerieDataset(X_tr, y_tr), batch_size=batch, shuffle=False)
    loader_teste  = DataLoader(SerieDataset(X_te, y_te), batch_size=batch, shuffle=False)

    print(f'Treino: {len(X_tr):>5} janelas  ({len(X_tr)//len(tickers)} por ativo)')
    print(f'Teste:  {len(X_te):>5} janelas  ({len(X_te)//len(tickers)} por ativo)')

    return loader_treino, loader_teste, janelas, scalers


loader_treino, loader_teste, janelas, scalers = preparar_pipeline(
    features_ativos, TICKERS_ATIVOS, LOOKBACK, BATCH
)
````

## 5. Arquitetura LSTM

```
Entrada (batch, 20, 13)  →  LSTM camada 1 (hidden=128, dropout=0.2)
                          →  LSTM camada 2 (hidden=128, dropout=0.2)
                          →  Linear(128 → 1)
```

**Decisões de projeto e suas justificativas:**

- **Saída linear (sem ativação):** log-retornos assumem valores em ℝ sem limite superior ou inferior. Funções como `tanh` ou `sigmoid` restringiriam artificialmente o espaço de saída e introduziriam viés sistemático em retornos de maior magnitude.
- **2 camadas LSTM:** a hierarquia de camadas permite capturar dependências temporais em múltiplas escalas — padrões de curto prazo na camada 1 e relações de prazo mais longo na camada 2.
- **Dropout = 0,2:** aplicado entre as camadas recorrentes como mecanismo de regularização. Em séries temporais financeiras, dropout em LSTMs é discutido na literatura (Gal & Ghahramani, 2016); aqui é usado como heurística padrão, sem ablation formal.

**Função de perda híbrida:**

$$\mathcal{L} = \text{RMSE}(y, \hat{y}) + \lambda \cdot \left(1 - \frac{\text{Sharpe}_{\text{estratégia}} + 2}{4}\right), \quad \lambda = 0.3$$

O termo RMSE penaliza o erro quadrático de previsão. O termo Sharpe normalizado alinha o objetivo de otimização com o resultado financeiro final — um modelo que minimiza apenas o RMSE pode gerar previsões estatisticamente acuradas mas com direção sistematicamente incorreta.

> **Transparência sobre λ:** o valor λ=0,3 foi escolhido empiricamente, sem busca sistemática em grid. Valores de λ entre 0,1 e 0,5 produzem comportamentos similares nos experimentos exploratórios, mas uma análise de sensibilidade formal não foi realizada e constitui limitação reconhecida.


````python
class EnergyLSTM(nn.Module):
    def __init__(self, input_size=13, hidden_size=128, num_layers=2, dropout=0.2):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size, hidden_size, num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0.0
        )
        self.fc = nn.Linear(hidden_size, 1)  # sem ativação — regressão

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1, :])


def sharpe_loss(y_pred, y_true, lamb=0.3, eps=1e-8):
    rmse     = torch.sqrt(torch.mean((y_pred.squeeze() - y_true) ** 2))
    retornos = y_true * torch.sign(y_pred.squeeze())
    sharpe_n = torch.tanh(retornos.mean() / (retornos.std() + eps))
    return rmse + lamb * (1.0 - sharpe_n)


modelo = EnergyLSTM(input_size=len(FEATURES)).to(DEVICE)
total  = sum(p.numel() for p in modelo.parameters() if p.requires_grad)
print(modelo)
print(f'\nParâmetros treináveis: {total:,}')
````

## 6. Treinamento com Early Stopping

**Hiperparâmetros:**

| Parâmetro | Valor |
|-----------|-------|
| Otimizador | Adam, lr=1e-3 |
| Scheduler | ReduceLROnPlateau (patience=5, fator=0,5) |
| Early stopping | patience=10, máx. 100 épocas |
| Gradient clipping | max_norm=1,0 |
| Batch size | 64 |

**Limitação do protocolo de parada antecipada:** o early stopping nesta implementação monitora a loss no **conjunto de treino**, não em um conjunto de validação separado. Isso implica que o critério de parada serve apenas para limitar o número de épocas, sem garantir que o modelo não esteja sobreajustando aos dados de treino. A ausência de um conjunto de validação independente é uma fragilidade metodológica reconhecida; a análise de robustez com 50 seeds (Seção 13) fornece evidência indireta contra overfitting severo, mas não substitui a validação formal.

O gradient clipping (`max_norm=1,0`) previne o problema de *exploding gradients*, recorrente em LSTMs treinadas em séries com alta volatilidade.


````python
LR            = 1e-3
EPOCHS        = 100
PATIENCE      = 10
LAMBDA_SHARPE = 0.3

otimizador = torch.optim.Adam(modelo.parameters(), lr=LR)
scheduler  = torch.optim.lr_scheduler.ReduceLROnPlateau(
    otimizador, mode='min', patience=5, factor=0.5
)

def treinar_epoca(modelo, loader):
    modelo.train()
    total = 0.0
    for X_b, y_b in loader:
        X_b, y_b = X_b.to(DEVICE), y_b.to(DEVICE)
        otimizador.zero_grad()
        loss = sharpe_loss(modelo(X_b), y_b, LAMBDA_SHARPE)
        loss.backward()
        nn.utils.clip_grad_norm_(modelo.parameters(), 1.0)
        otimizador.step()
        total += loss.item()
    return total / len(loader)


historico    = []
melhor_loss  = float('inf')
melhor_pesos = None
paciencia    = 0

for epoch in range(1, EPOCHS + 1):
    loss = treinar_epoca(modelo, loader_treino)
    historico.append(loss)
    scheduler.step(loss)

    if loss < melhor_loss:
        melhor_loss  = loss
        melhor_pesos = {k: v.clone() for k, v in modelo.state_dict().items()}
        paciencia    = 0
    else:
        paciencia += 1

    if epoch % 10 == 0 or epoch == 1:
        lr_atual = otimizador.param_groups[0]['lr']
        print(f'Época {epoch:>3}  loss={loss:.5f}  lr={lr_atual:.2e}  paciência={paciencia}/{PATIENCE}')

    if paciencia >= PATIENCE:
        print(f'\nEarly stopping na época {epoch}. Melhor loss: {melhor_loss:.5f}')
        break

modelo.load_state_dict(melhor_pesos)
print('Melhores pesos restaurados.')
````

````python
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(historico, color='#1D9E75', linewidth=1.5)
ax.axvline(np.argmin(historico), color='#888', linestyle='--',
           linewidth=0.9, label=f'Mínimo (época {np.argmin(historico)+1})')
ax.set_xlabel('Época')
ax.set_ylabel('Loss (híbrida)')
ax.set_title('Curva de Aprendizado — Treino', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.savefig('curva_aprendizado.png', dpi=150, bbox_inches='tight')
plt.show()
````

## 7. Threshold Dinâmico — Calibração no Conjunto de Treino

**Motivação:** um threshold fixo ignora a heterogeneidade de escala entre ativos. HAL possui volatilidade historicamente superior a XOM; um threshold único seria simultaneamente permissivo para um e restritivo para o outro.

**Procedimento:**

1. Calcular as previsões `|ŷ|` no conjunto de treino para cada ativo
2. Definir o threshold como o percentil `p` dessa distribuição
3. Testar múltiplos percentis e selecionar o que **maximiza o Sharpe anualizado no treino**
4. Aplicar o threshold selecionado no teste **sem reajuste**

**Regra de operação:**
- `|ŷ| > threshold` e `ŷ > 0` → posição *long*
- `|ŷ| > threshold` e `ŷ < 0` → posição *short*
- `|ŷ| ≤ threshold` → cash (proteção de capital)

> **Risco de data snooping:** a seleção do percentil ótimo por maximização do Sharpe no treino é uma forma de otimização em amostra. Embora o threshold seja aplicado sem reajuste no teste, o fato de o critério de seleção ser o Sharpe — e não uma métrica de previsão como RMSE ou acurácia direcional — introduz um viés implícito de alinhamento com o resultado financeiro. Este procedimento é metodologicamente mais fraco do que fixar o percentil *a priori* com base em critério econômico independente.


````python
def prever(modelo, X_np):
    modelo.eval()
    with torch.no_grad():
        return modelo(torch.from_numpy(X_np).to(DEVICE)).cpu().numpy().squeeze()

def retornos_estrategia(y_true, y_pred, threshold):
    sinais = np.where(np.abs(y_pred) > threshold, np.sign(y_pred), 0.0)
    return sinais * y_true

def sharpe_anualizado(retornos, dias_ano=252):
    std = retornos.std()
    return (retornos.mean() / std * np.sqrt(dias_ano)) if std > 1e-10 else 0.0

def max_drawdown(curva_capital):
    pico = np.maximum.accumulate(curva_capital)
    return ((curva_capital - pico) / pico).min()


# ── Previsões no treino (para calibração) ─────────────────────────────────────
previsoes_treino = {}
for t in TICKERS_ATIVOS:
    X_tr, y_tr = janelas['treino'][t]
    previsoes_treino[t] = {'y_pred': prever(modelo, X_tr), 'y_true': y_tr}

# ── Busca do percentil ótimo ───────────────────────────────────────────────────
# Percentis candidatos — começamos em p30 para garantir que pelo menos
# 30% dos dias fiquem em cash. Percentis baixos (p0, p10) colapsam para
# thresholds na casa de 0.000001 porque a distribuição de |ŷ| é muito
# concentrada perto de zero — o filtro se torna inefetivo.
PERCENTIS = [30, 40, 50, 60, 70, 80]
# p30 → opera os 70% de dias com maior |ŷ|
# p50 → opera os 50% de dias com maior |ŷ|
# p80 → opera apenas os 20% de dias com maior convicção

print('Calibração do threshold dinâmico — Sharpe médio no treino:')
print(f'  {"Percentil":<12} {"Threshold médio":<18} {"Sharpe médio":<15} {"% operando"}')
print('  ' + '-'*58)

sharpes_por_percentil = {}
thresholds_por_ativo  = {}   # threshold individual por ativo

for pct in PERCENTIS:
    sharpes    = []
    thresholds = []
    cobert     = []
    for t in TICKERS_ATIVOS:
        yp  = previsoes_treino[t]['y_pred']
        yt  = previsoes_treino[t]['y_true']
        thr = np.percentile(np.abs(yp), pct)
        ret = retornos_estrategia(yt, yp, thr)
        sharpes.append(sharpe_anualizado(ret))
        thresholds.append(thr)
        cobert.append(np.mean(np.abs(yp) > thr))
    media_sharpe = np.mean(sharpes)
    media_thr    = np.mean(thresholds)
    media_cob    = np.mean(cobert)
    sharpes_por_percentil[pct] = media_sharpe
    print(f'  p{pct:<11} {media_thr:<18.5f} {media_sharpe:<15.3f} {media_cob:.1%}')

PERCENTIL_OTIMO = max(sharpes_por_percentil, key=sharpes_por_percentil.get)
print(f'\nPercentil ótimo: p{PERCENTIL_OTIMO}')

# Calcula threshold individual por ativo com o percentil ótimo
for t in TICKERS_ATIVOS:
    yp = previsoes_treino[t]['y_pred']
    thresholds_por_ativo[t] = np.percentile(np.abs(yp), PERCENTIL_OTIMO)
    print(f'  Threshold {t}: {thresholds_por_ativo[t]:.6f}')
````

````python
# Visualização: Sharpe vs percentil de threshold
fig, ax = plt.subplots(figsize=(9, 4))
sharpe_vals = [sharpes_por_percentil[p] for p in PERCENTIS]
cores_barras = ['#1D9E75' if p == PERCENTIL_OTIMO else '#cccccc' for p in PERCENTIS]
ax.bar([f'p{p}' for p in PERCENTIS], sharpe_vals, color=cores_barras, edgecolor='white')
ax.axhline(0, color='#888', linewidth=0.7)
ax.set_xlabel('Percentil do threshold')
ax.set_ylabel('Sharpe anualizado (treino)')
ax.set_title('Calibração do Threshold Dinâmico\n'
             'p30 = opera 70% dos dias | p80 = opera 20% dos dias (maior convicção)',
             fontweight='bold')
plt.tight_layout()
plt.savefig('calibracao_threshold.png', dpi=150, bbox_inches='tight')
plt.show()
````

## 8. Avaliação Estatística no Teste (2023–2025)

Duas categorias de métricas são reportadas separadamente, pois avaliam dimensões distintas do modelo:

**Métricas de qualidade da previsão** (avaliam se o modelo estima corretamente a magnitude e direção dos retornos):

| Métrica | Fórmula | Referência para finanças |
|---------|---------|--------------------------|
| RMSE | √mean((ŷ − y)²) | Sensível a outliers; útil para comparar modelos |
| MAE | mean(|ŷ − y|) | Mais robusto a outliers que o RMSE |
| R² | 1 − SS_res/SS_tot | **0,01–0,05 é estado-da-arte** (Gu et al., 2020) |
| Acurácia direcional | mean(sign(ŷ) == sign(y)) | Acima de ~52% é economicamente significativo |

### O debate do R² — por que valores baixos são esperados e o que isso significa

**Perspectiva A — majoritária na literatura de ML financeiro:**  
R² baixo é *esperado e normal* em previsão de retornos diários. Preços diários incorporam informação nova de forma quase imediata (hipótese semi-forte de mercados eficientes, Fama 1970) — por construção, a relação sinal/ruído é próxima de zero. Gu, Kelly & Xiu (2020) reportam R²_oos de 0,4–1,4% como estado-da-arte usando ensembles com milhares de ações americanas. Valores de 1–3% já são considerados "economicamente significativos" nessa literatura. O que importa é se o modelo acerta a *direção* com frequência suficiente para gerar alpha — não se estima a *magnitude* com precisão.

**Perspectiva B — crítica legítima:**  
Um R² muito próximo de zero pode indicar que o modelo aprendeu apenas o viés da média dos retornos de treino, sem capturar sinal real. Nesse caso, acurácia direcional acima de 50% poderia ser artefato da assimetria da distribuição de retornos no período de teste (e.g., mercado em tendência de alta). O teste correto é verificar se a acurácia direcional é significativamente superior a 50% por **teste binomial bilateral** — e não apenas observá-la descritivamente.

**Posição analítica adotada:** R² abaixo de 0,05 não invalida a hipótese, desde que a acurácia direcional seja estatisticamente superior a 50% e o alpha seja significativo por bootstrap. A evidência que resolve o debate não é o R² em si — é o resultado dos testes de significância abaixo.

### Testes de significância implementados

1. **Teste binomial bilateral** na acurácia direcional: H₀ = acurácia = 0,50
2. **Bootstrap (1000 iterações)** no Sharpe anualizado: IC 95% por percentil
3. **Teste t de Student** no alpha diário médio: H₀ = mean(alpha_diário) = 0

Esses três testes juntos respondem à questão central: *o resultado é sinal ou ruído?*

**Interpretação:** métricas estatísticas fracas (baixo R², acurácia direcional próxima a 50% não-significativa) não invalidam automaticamente a estratégia — mas exigem que o alpha financeiro seja sustentado pelos testes acima. Por outro lado, performance financeira positiva com métricas estatísticas ruins levanta a questão de se o resultado é produto do threshold e não da qualidade da previsão em si.


````python
from scipy.stats import binomtest, ttest_1samp

def acuracia_direcional(y_true, y_pred):
    return np.mean(np.sign(y_true) == np.sign(y_pred))

def bootstrap_sharpe(retornos, n_boot=1000, seed=0):
    """IC 95% do Sharpe anualizado por bootstrap percentil."""
    rng = np.random.default_rng(seed)
    sharpes = []
    for _ in range(n_boot):
        s = rng.choice(retornos, size=len(retornos), replace=True)
        sharpes.append(sharpe_anualizado(s))
    return np.percentile(sharpes, [2.5, 97.5])

previsoes_teste = {}
metricas_estat  = {}

for t in TICKERS_ATIVOS:
    X_te, y_te = janelas['teste'][t]
    y_pred = prever(modelo, X_te)
    previsoes_teste[t] = {'y_true': y_te, 'y_pred': y_pred}

    rmse = np.sqrt(mean_squared_error(y_te, y_pred))
    mae  = mean_absolute_error(y_te, y_pred)
    r2   = r2_score(y_te, y_pred)
    dacc = acuracia_direcional(y_te, y_pred)

    # ── Teste binomial bilateral: H0 = acc = 0.5 ─────────────────────────────
    n_total   = len(y_te)
    n_acertos = int(np.sum(np.sign(y_te) == np.sign(y_pred)))
    binom_res = binomtest(n_acertos, n_total, p=0.5, alternative='two-sided')
    dacc_pval = binom_res.pvalue

    # ── Acurácia direcional apenas nos dias que passam pelo threshold ─────────
    thr  = thresholds_por_ativo[t]
    mask = np.abs(y_pred) > thr
    n_f   = int(mask.sum())
    dacc_f = acuracia_direcional(y_te[mask], y_pred[mask]) if n_f > 0 else np.nan
    n_ac_f = int(np.sum(np.sign(y_te[mask]) == np.sign(y_pred[mask]))) if n_f > 0 else 0
    binom_f = binomtest(n_ac_f, n_f, p=0.5, alternative='two-sided') if n_f > 0 else None
    dacc_f_pval = binom_f.pvalue if binom_f else np.nan

    metricas_estat[t] = {
        'RMSE': rmse, 'MAE': mae, 'R²': r2,
        'Dir.Acc. (todos)': dacc,
        'p-valor (binomial)': dacc_pval,
        'Dir.Acc. (filtrado)': dacc_f,
        'p-valor filtrado': dacc_f_pval,
        'Dias filtrados': n_f,
    }

df_estat = pd.DataFrame(metricas_estat).T.round(4)
print('=== Métricas Estatísticas — Teste 2023–2025 ===')
print(df_estat.to_string())

# ── Interpretação do R² ───────────────────────────────────────────────────────
print()
print('=== Contextualização do R² (referência: Gu et al., 2020) ===')
for t in TICKERS_ATIVOS:
    r2 = metricas_estat[t]['R²']
    if r2 >= 0.02:
        interp = 'Economicamente significativo (≥ 0,02)'
    elif r2 >= 0.005:
        interp = 'Na faixa reportada em literatura (0,005–0,02)'
    else:
        interp = 'Abaixo do típico — sinal pode ser dominado por ruído'
    print(f'  {t}: R²={r2:.4f}  → {interp}')

# ── Significância da acurácia direcional ──────────────────────────────────────
print()
print('=== Significância — Acurácia Direcional Filtrada ===')
print('  H₀: acurácia = 50%  |  Teste binomial bilateral  |  α = 0,05')
for t in TICKERS_ATIVOS:
    pv  = metricas_estat[t]['p-valor filtrado']
    acc = metricas_estat[t]['Dir.Acc. (filtrado)']
    sig = '✓ SIGNIFICATIVO' if pv < 0.05 else '✗ não significativo'
    print(f'  {t}: acc={acc:.3f}  p={pv:.4f}  → {sig}')
````

````python
# ── Bootstrap IC 95% do Sharpe e t-test do alpha diário ─────────────────────
# Esses dois testes são complementares:
# - Bootstrap do Sharpe: não assume normalidade dos retornos (mais robusto)
# - t-test do alpha diário: paramétrico; detecta se o retorno médio é diferente de zero
# Um resultado consistente entre os dois fortalece a conclusão.

print('=== Bootstrap IC 95% do Sharpe (1000 amostras) + t-test do alpha ===')
print(f'  {"Ativo":<6}  {"Sharpe":>8}  {"IC 95% inferior":>16}  {"IC 95% superior":>16}  {"t-test p":>10}  {"Sinal?"}')
print('  ' + '─' * 78)

for t in TICKERS_ATIVOS:
    yt  = previsoes_teste[t]['y_true']
    yp  = previsoes_teste[t]['y_pred']
    thr = thresholds_por_ativo[t]
    ret = retornos_estrategia(yt, yp, thr)

    sharpe_pt = sharpe_anualizado(ret)
    ic        = bootstrap_sharpe(ret)

    # t-test apenas nos dias operados (ret != 0)
    ret_op = ret[ret != 0]
    _, pval = ttest_1samp(ret_op, 0) if len(ret_op) > 1 else (np.nan, np.nan)

    # IC que não cruza zero E t-test significativo = evidência mais forte
    ic_positivo = ic[0] > 0
    ttest_sig   = pval < 0.05 if not np.isnan(pval) else False
    sinal = '✓✓ forte' if (ic_positivo and ttest_sig) else ('✓ parcial' if (ic_positivo or ttest_sig) else '✗ sem sinal')

    print(f'  {t:<6}  {sharpe_pt:>+8.3f}  {ic[0]:>+16.3f}  {ic[1]:>+16.3f}  {pval:>10.4f}  {sinal}')

print()
print('Legenda:')
print('  ✓✓ forte   → IC não cruza zero E t-test significativo (p < 0,05)')
print('  ✓ parcial  → apenas um dos dois critérios satisfeito')
print('  ✗ sem sinal → nenhum critério satisfeito; alpha pode ser ruído')
````

## 9. Estratégia de Trading e Simulação

**Regra de decisão:**
- `ŷ > threshold` → posição *long* (compra)
- `ŷ < −threshold` → posição *short* (venda a descoberto)
- `|ŷ| ≤ threshold` → *cash* (sem exposição)

Threshold individual por ativo, calibrado no percentil ótimo da distribuição de treino (Seção 7).

**Premissas e limitações explícitas da simulação:**

| Premissa | Impacto esperado |
|----------|-----------------|
| Sem custos de transação (bid-ask ~0,05% por operação) | Superestima o retorno realizado; em estratégia diária, o custo acumulado pode eliminar parte relevante do alpha |
| Short selling sem custo de aluguel (*borrow rate*) | Superestima retornos em períodos de alta demanda por posições vendidas |
| Execução ao preço de fechamento do dia *t* | Implica olhar para frente (*look-ahead*) no dia de geração do sinal; na prática, a execução ocorre no dia *t+1* |
| Sem restrição de liquidez ou impacto de mercado | Válida para os ativos selecionados (alta liquidez), mas limita generalizabilidade para small/mid caps |

> O retorno reportado deve ser interpretado como **limite superior teórico** do alpha realizável, não como estimativa do retorno líquido.


````python
def calcular_metricas_trading(retornos, nome, tipo):
    curva     = np.exp(np.cumsum(retornos))
    ret_total = curva[-1] - 1
    sharpe    = sharpe_anualizado(retornos)
    mdd       = max_drawdown(curva)
    dias_op   = int(np.sum(retornos != 0))
    pct_op    = dias_op / len(retornos)
    return {
        'Nome': nome, 'Tipo': tipo,
        'Retorno total': ret_total,
        'Sharpe':        sharpe,
        'Max Drawdown':  mdd,
        'Dias operando': dias_op,
        '% operando':    pct_op,
    }, curva


resultados = []
curvas     = {}

for t in TICKERS_ATIVOS:
    yt   = previsoes_teste[t]['y_true']
    yp   = previsoes_teste[t]['y_pred']
    thr  = thresholds_por_ativo[t]

    # LSTM com threshold dinâmico
    ret_lstm = retornos_estrategia(yt, yp, thr)
    m, c     = calcular_metricas_trading(ret_lstm, t, 'LSTM')
    m['Dir.Acc.'] = metricas_estat[t]['Dir.Acc. (filtrado)']
    resultados.append(m)
    curvas[f'LSTM_{t}'] = c

    # Buy & hold
    ret_bh   = yt
    m_bh, c_bh = calcular_metricas_trading(ret_bh, t, 'B&H')
    m_bh['Dir.Acc.'] = np.nan
    resultados.append(m_bh)
    curvas[f'BH_{t}'] = c_bh

# Carteira equal-weight
ret_lstm_cart = np.mean(
    [retornos_estrategia(previsoes_teste[t]['y_true'],
                         previsoes_teste[t]['y_pred'],
                         thresholds_por_ativo[t])
     for t in TICKERS_ATIVOS], axis=0)
ret_bh_cart = np.mean([previsoes_teste[t]['y_true'] for t in TICKERS_ATIVOS], axis=0)

m_lc, c_lc = calcular_metricas_trading(ret_lstm_cart, 'Carteira', 'LSTM')
m_lc['Dir.Acc.'] = np.nan
resultados.append(m_lc)
curvas['LSTM_Carteira'] = c_lc

m_bc, c_bc = calcular_metricas_trading(ret_bh_cart, 'Carteira', 'B&H')
m_bc['Dir.Acc.'] = np.nan
resultados.append(m_bc)
curvas['BH_Carteira'] = c_bc

df_res = pd.DataFrame(resultados)

# Formata para exibição
df_print = df_res.copy()
df_print['Retorno total'] = df_print['Retorno total'].map('{:.2%}'.format)
df_print['Sharpe']        = df_print['Sharpe'].round(3)
df_print['Max Drawdown']  = df_print['Max Drawdown'].map('{:.2%}'.format)
df_print['% operando']    = df_print['% operando'].map('{:.1%}'.format)
df_print['Dir.Acc.']      = df_print['Dir.Acc.'].apply(
    lambda x: f'{x:.2%}' if pd.notna(x) else '—')

cols = ['Nome', 'Tipo', 'Retorno total', 'Sharpe', 'Max Drawdown',
        'Dias operando', '% operando', 'Dir.Acc.']
print('\n' + '='*80)
print('RESULTADOS — Teste 2023–2025')
print('='*80)
print(df_print[df_print['Nome'] != 'Carteira'][cols].to_string(index=False))
print('\n--- Carteira igual ponderada ---')
print(df_print[df_print['Nome'] == 'Carteira'][cols].to_string(index=False))
print(f'\nThreshold: percentil p{PERCENTIL_OTIMO} da distribuição de |ŷ| no treino')
print(f'Sharpe anualizado: 252 dias de pregão/ano')
````

## 12. Análise de Robustez — 50 Seeds Independentes

**Motivação:** modelos com pesos inicializados aleatoriamente podem apresentar alta variância em métricas de performance. Um único resultado favorável pode ser artefato de uma inicialização sortuda (*lucky seed*). A análise de robustez replica o pipeline completo — incluindo treinamento, calibração de threshold e simulação no teste — com 50 seeds independentes.

**Protocolo:**
- Seeds 0 a 49, todos os demais hiperparâmetros e splits idênticos
- Métricas coletadas: Sharpe da carteira, capital final, alpha vs. B&H, máximo drawdown
- CSV incremental: salvo após cada run; execução interrompida pode ser retomada automaticamente

**Interpretação estatística dos resultados:**
- A proporção de runs com alpha > 0 é reportada como estimativa empírica da consistência do modelo.
- O intervalo [média ± 1 desvio padrão] do alpha quantifica a incerteza associada à inicialização aleatória.
- Um desvio padrão elevado em relação à média indica que a performance é sensível a hiperparâmetros não controlados (pesos iniciais) — o que é esperado em modelos financeiros, mas deve ser reportado transparentemente.

> **Tempo estimado com GPU (Kaggle/Colab):** 15–30 min.


````python
import os, time

N_SEEDS  = 50
SEEDS    = list(range(N_SEEDS))
CSV_PATH = 'robustez_runs.csv'


def run_seed(seed):
    """Treina do zero com seed fixo e retorna métricas do teste."""
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    # Modelo e otimizador
    m   = EnergyLSTM(input_size=len(FEATURES)).to(DEVICE)
    opt = torch.optim.Adam(m.parameters(), lr=LR)
    sch = torch.optim.lr_scheduler.ReduceLROnPlateau(
        opt, mode='min', patience=5, factor=0.5
    )

    # Treino com early stopping
    best_loss, best_w, pac = float('inf'), None, 0
    for epoch in range(1, EPOCHS + 1):
        m.train()
        ep_loss = 0.0
        for Xb, yb in loader_treino:
            Xb, yb = Xb.to(DEVICE), yb.to(DEVICE)
            opt.zero_grad()
            loss = sharpe_loss(m(Xb), yb, LAMBDA_SHARPE)
            loss.backward()
            nn.utils.clip_grad_norm_(m.parameters(), 1.0)
            opt.step()
            ep_loss += loss.item()
        ep_loss /= len(loader_treino)
        sch.step(ep_loss)
        if ep_loss < best_loss:
            best_loss = ep_loss
            best_w    = {k: v.clone() for k, v in m.state_dict().items()}
            pac       = 0
        else:
            pac += 1
        if pac >= PATIENCE:
            break
    m.load_state_dict(best_w)

    # Threshold dinâmico — mesmo percentil ótimo já calibrado
    thr = {}
    for t in TICKERS_ATIVOS:
        Xtr, _ = janelas['treino'][t]
        thr[t] = np.percentile(np.abs(prever(m, Xtr)), PERCENTIL_OTIMO)

    # Avaliação no teste
    cap_lstm      = np.zeros(len(curvas['LSTM_XOM']))
    cap_bh        = np.zeros(len(curvas['BH_XOM']))
    ret_lstm_list = []

    for t in TICKERS_ATIVOS:
        Xte, yte = janelas['teste'][t]
        ypte     = prever(m, Xte)
        ret_l    = retornos_estrategia(yte, ypte, thr[t])
        cap_lstm += simular_capital(ret_l, CAPITAL_POR_ATIVO)
        cap_bh   += simular_capital(yte,   CAPITAL_POR_ATIVO)
        ret_lstm_list.append(ret_l)

    ret_cart = np.mean(ret_lstm_list, axis=0)

    return {
        'seed':         seed,
        'epocas':       epoch,
        'sharpe_cart':  sharpe_anualizado(ret_cart),
        'capital_lstm': cap_lstm[-1],
        'capital_bh':   cap_bh[-1],
        'alpha_usd':    cap_lstm[-1] - cap_bh[-1],
        'mdd_cart':     max_drawdown(np.exp(np.cumsum(ret_cart))),
        'lstm_ganhou':  int(cap_lstm[-1] > cap_bh[-1]),
    }


# Carrega runs anteriores se o CSV já existir
if os.path.exists(CSV_PATH):
    df_runs      = pd.read_csv(CSV_PATH)
    seeds_feitos = set(df_runs['seed'].tolist())
    print(f'CSV encontrado — {len(df_runs)} runs já salvos. Continuando...')
else:
    df_runs      = pd.DataFrame()
    seeds_feitos = set()
    print('Iniciando do zero — 50 runs')

pendentes = [s for s in SEEDS if s not in seeds_feitos]
print(f'Seeds pendentes: {len(pendentes)}\n')

for i, seed in enumerate(pendentes, start=1):
    t0  = time.time()
    res = run_seed(seed)
    dt  = time.time() - t0

    df_runs = pd.concat([df_runs, pd.DataFrame([res])], ignore_index=True)
    df_runs.to_csv(CSV_PATH, index=False)

    done   = len(seeds_feitos) + i
    status = '✓' if res['lstm_ganhou'] else '✗'
    print(f'[{done:>2}/50]  seed={seed:>2}  '
          f'sharpe={res["sharpe_cart"]:>+.3f}  '
          f'alpha=${res["alpha_usd"]:>+.0f}  '
          f'capital=${res["capital_lstm"]:>,.0f}  '
          f'epochs={res["epocas"]:>3}  {dt:.0f}s  {status}')

print(f'\nConcluído. CSV salvo em: {CSV_PATH}')
````
